In a 1D linear finite element, we can begin by writing the approximate solution as a straight line:

$$
u(x) = a_0 + a_1 x
$$

Here, $a_0$ and $a_1$ are simply the coefficients of that line:

- $a_0$ is the intercept,
- $a_1$ is the slope.

To make this line match the nodal values, we impose

$$
u_1 = u(x_1), \qquad u_2 = u(x_2)
$$

so that

$$
u_1 = a_0 + a_1 x_1,
\qquad
u_2 = a_0 + a_1 x_2
$$

These two equations allow us to solve for $a_0$ and $a_1$ in terms of the nodal values $u_1$ and $u_2$:

$$
a_0 = \frac{u_1 x_2 - u_2 x_1}{x_2-x_1},
\qquad
a_1 = \frac{u_2-u_1}{x_2-x_1}
$$

If we substitute these expressions back into

$$
u(x) = a_0 + a_1 x
$$

and rearrange the terms, we obtain

$$
u(x) = N_1(x)u_1 + N_2(x)u_2
$$

where

$$
N_1(x) = \frac{x_2-x}{x_2-x_1},
\qquad
N_2(x) = \frac{x-x_1}{x_2-x_1}
$$

So the relation is that $N_1$ and $N_2$ come from eliminating $a_0$ and $a_1$ and rewriting the same straight line in terms of the nodal values.

This means that the two forms

$$
u(x)=a_0+a_1x
\qquad \text{and} \qquad
u(x)=N_1(x)u_1+N_2(x)u_2
$$

represent exactly the same approximation. The difference is only in how the approximation is written.

The form with $a_0$ and $a_1$ is the usual equation of a line. The form with $N_1$ and $N_2$ is the finite element form, because it expresses the solution directly in terms of the nodal unknowns.

The functions $N_1$ and $N_2$ are called **shape functions** or **interpolation functions** because they tell us how much each node contributes to the value of $u(x)$ at any point inside the element.

They also satisfy the nodal properties

$$
N_1(x_1)=1,\quad N_1(x_2)=0,
\qquad
N_2(x_1)=0,\quad N_2(x_2)=1
$$

so each shape function is equal to 1 at its own node and 0 at the other node.

In summary:

- $a_0$ and $a_1$ define the straight line in standard algebraic form.
- $N_1$ and $N_2$ define the same line in nodal interpolation form.
- The shape functions are obtained by rewriting the line so that the nodal values appear explicitly.
- This FEM form is more convenient because the unknowns are the nodal values $u_1$ and $u_2$, not the coefficients $a_0$ and $a_1$.

In the finite element method, we approximate the unknown function inside each element with a simple polynomial. For a 1D linear element, the approximation is

$$
u(x) = N_1(x)u_1 + N_2(x)u_2
$$

where $u_1$ and $u_2$ are the nodal values, and $N_1(x)$ and $N_2(x)$ are the linear shape functions.

For a linear element,

$$
N_1(x) = \frac{x_2-x}{x_2-x_1},
\qquad
N_2(x) = \frac{x-x_1}{x_2-x_1}
$$

These are first-degree polynomials, so their derivatives are very simple:

$$
\frac{dN_1}{dx} = -\frac{1}{x_2-x_1},
\qquad
\frac{dN_2}{dx} = \frac{1}{x_2-x_1}
$$

Therefore, differentiating the approximate solution is straightforward:

$$
\frac{du}{dx}
=
\frac{dN_1}{dx}u_1 + \frac{dN_2}{dx}u_2
$$

Since the derivatives of the shape functions are constants, the slope inside the element is easy to compute.

Integration is also simple. For example,

$$
\int u \, dx
=
\int \left(N_1u_1 + N_2u_2\right)\,dx
$$

Because $u_1$ and $u_2$ are constants within the element, we can write

$$
\int u \, dx
=
u_1 \int N_1\,dx + u_2 \int N_2\,dx
$$

So the integral reduces to the integral of simple linear functions.

This is why linear equations facilitate differentiation and integration: by using a linear approximation within each element, the calculus becomes easy and can often be done analytically.

A useful teaching summary is:

- In a linear element, the solution is approximated by a straight line.
- The derivative of a straight line is a constant.
- The integral of a straight line is simple to evaluate.
- This makes the element equations much easier to construct.

So, linear finite elements are popular not only because they are simple conceptually, but also because they make the mathematical operations needed in FEM much easier.

In the finite element method, there are several ways to derive the element equations, including the direct approach, the method of weighted residuals, and the variational approach. Although these methods start from different ideas, they often lead to the same final result: a system of linear algebraic equations written in matrix form as

$$
[k]\{u\} = \{F\}
$$

This equation is the finite element version of the governing differential equation over a single element.

Here,

$$
[k]
$$

is the **element stiffness matrix** (or, more generally, the element property matrix). It contains the coefficients that describe how the element resists deformation or transfer. Its entries depend on the material properties, the geometry of the element, and the choice of shape functions.

The vector

$$
\{u\}
$$

is the **column vector of nodal unknowns**. These are the values we want to solve for, such as displacement, temperature, or pressure, depending on the physical problem.

The vector

$$
\{F\}
$$

is the **element force vector** (or load vector). It represents the effect of external influences applied to the element, such as forces, distributed loads, heat sources, or boundary contributions.

- The differential equation describes the physics continuously over the domain.
- The finite element method replaces that continuous problem with a set of algebraic equations.
- Each element contributes one matrix equation of the form
  $$
  [k]\{u\} = \{F\}
  $$
- After all elements are assembled, the full system for the whole domain is obtained.

So this matrix equation is important because it shows the bridge between the original calculus-based problem and the final linear algebra problem that we solve on a computer.

A simple interpretation is:

- $[k]$ tells us how the element behaves,
- $\{u\}$ contains the unknown nodal values,
- $\{F\}$ represents the applied effects.

Thus, finite elements convert a boundary-value problem into a matrix problem.


> In FEM, the governing differential equation is transformed into a matrix equation for each element. The stiffness matrix contains the physical and geometric properties of the element, the unknown vector contains the nodal values, and the force vector represents the external actions. Once all element equations are assembled, we solve a global linear system for the entire structure or domain.

## Assembly

Assembly is the step in the finite element method where the equations for the individual elements are combined into one larger system for the entire domain.

The key idea behind assembly is **continuity**. Neighboring elements share nodes, so the solution must match at those common nodes. In other words, if two elements meet at the same node, they must have the same value of the unknown there. In some problems, certain derivatives may also need to be continuous, but for basic 1D finite elements the main requirement is continuity of the unknown itself.

Each element has its own matrix equation of the form

$$
[k]\{u\} = \{F\}
$$

where the matrices and vectors apply only to that single element.

During assembly, all of the element equations are merged into a global system:

$$
[K]\{u'\} = \{F'\}
$$

Here,

$$
[K]
$$

is the **global stiffness matrix** or **assembled property matrix**. It is formed by combining the contributions of all the individual element matrices.

The vector

$$
\{u'\}
$$

is the **global vector of unknown nodal values** for the entire domain.

The vector

$$
\{F'\}
$$

is the **global force vector** or **assembled load vector**, which contains the combined effect of all external influences acting on the full system.

- Each element gives a small local equation.
- Neighboring elements share nodes.
- Because shared nodes must have the same unknown value, the local equations are connected together.
- This produces one global matrix system for the whole problem.

So assembly is the process that turns many small element equations into one large equation that describes the behavior of the complete structure or domain.

A simple physical interpretation is that each element contributes part of the stiffness and part of the loading, and the assembly process adds those contributions at the appropriate shared nodes.

> In FEM, we first write equations for each element separately. Then we assemble them by matching the shared nodes between adjacent elements. This creates the global system of equations, which can then be solved for the nodal unknowns over the entire domain.

## Test functions
In the weak form, the **test functions** are not automatically the same as the functions used to approximate the unknown, but in the most common finite element formulation they are chosen to be the same.

If we approximate the solution as

$$
u(x) \approx \sum_{j=1}^{n} N_j(x)\,u_j
$$

then the functions $N_j(x)$ are the **shape functions** or **hat functions**. They are used to interpolate the variable between the nodes.

In the weak form, we also introduce a function, often called $w(x)$, that multiplies the differential equation before integration. This is the **test function** or **weighting function**.

In the standard **Galerkin finite element method**, we choose the test functions from the same family as the shape functions. That is, for each equation we take

$$
w(x) = N_i(x)
$$

So in this very common case, the hat functions play two roles:

- they represent the approximate solution,
- and they are used as test functions in the weak form.

This is why, in many introductory FEM examples, the same functions appear in both places.

However, the two ideas are still conceptually different:

- **shape (trial) functions** are used to build the approximate solution,
- **test functions** are used to enforce the weak form.

They are equal in Galerkin FEM because we choose them to be equal, not because they must always be equal.

In other methods, such as **Petrov-Galerkin**, the test functions are different from the trial functions. So the safest summary is:

> In standard Galerkin finite elements, the test functions are the same as the hat functions used to interpolate the unknown, but this is a choice of method rather than a general rule.

## The weak form
Conceptually, the weak form is more forgiving than the strong form because it does not require the differential equation to be satisfied at every point. Instead, it requires the residual to satisfy an integral condition against all admissible test functions.

If the residual is

$$
R(x)
$$

then the weak form requires

$$
\int_{\Omega} w(x)\,R(x)\,dx = 0
\qquad \text{for all test functions } w(x)
$$

This means that the residual does not have to be zero everywhere. Instead, its **weighted average** must vanish for every test function. Because of this, positive and negative errors may cancel in an integral sense, but only in a way that makes the residual orthogonal to the whole test space.

This is why the weak form is called "weak": it enforces the governing equation in an averaged sense rather than point by point.

The test functions can also be viewed as **probes** of the residual. Each test function checks whether the residual has a nonzero weighted average over the region where that test function is active.

With low-order finite elements, such as piecewise linear hat functions, these probes are relatively smooth and local. As a result, the method is often more sensitive to broad, slowly varying errors than to very rapid oscillations.

So, in an approximate sense, the test functions act a bit like filters. They do not literally perform a Fourier low-pass filtering, but they make the method less sensitive to variations that cannot be well represented by the chosen finite element space.

This is why coarse meshes and low-order basis functions usually capture the large-scale behavior of the solution first, while finer details and shorter-wavelength variations require a finer mesh or higher-order elements.

A good summary is:

- The weak form enforces the equation in an averaged, weighted sense.
- Errors can cancel in the integrals, but only in a way consistent with all test functions.
- The test functions act like probes of the residual.
- Low-order test spaces are less sensitive to rapid variations than to smooth, large-scale behavior.

So the weak form is more forgiving, and the choice of test functions strongly influences which parts of the residual are seen most clearly.

## Integration by parts
A common step in deriving the weak form is **integration by parts**.  
For a 1D second-order equation,

$$
-\frac{d}{dx}\left(k\frac{du}{dx}\right)=f
$$

we first multiply by a test function $w(x)$ and integrate:

$$
\int_{\Omega} w\left(-\frac{d}{dx}\left(k\frac{du}{dx}\right)\right)\,dx
=
\int_{\Omega} wf\,dx
$$

Now apply integration by parts:

$$
\int_a^b w\,\frac{dv}{dx}\,dx
=
\left[w v\right]_a^b
-
\int_a^b \frac{dw}{dx}\,v\,dx
$$

with

$$
v = k\frac{du}{dx}
$$

Then,

$$
\int_{\Omega} w\frac{d}{dx}\left(k\frac{du}{dx}\right)\,dx
=
\left[w\,k\frac{du}{dx}\right]_{\partial\Omega}
-
\int_{\Omega} \frac{dw}{dx}\,k\frac{du}{dx}\,dx
$$

Therefore,

$$
\int_{\Omega} \frac{dw}{dx}\,k\frac{du}{dx}\,dx
=
\int_{\Omega} wf\,dx
+
\left[w\,k\frac{du}{dx}\right]_{\partial\Omega}
$$

This is the weak form. The important idea is that integration by parts reduces the order of the derivative on $u$ and introduces a boundary term.
